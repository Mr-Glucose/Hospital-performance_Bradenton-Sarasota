## 🛠️ Helper Functions

```python
def clean_measure_name(measure_id):
    """
    Convert measure codes to human-readable labels.
    
    Examples:
        H_COMP_1_A_P -> Communication with nurses (always)
        H_CLEAN_HSP_A_P -> Cleanliness of hospital environment (always)
    """
    mapping = {
        'H_COMP_1': 'Communication with nurses',
        'H_COMP_2': 'Communication with doctors',
        'H_COMP_3': 'Responsiveness of hospital staff',
        'H_COMP_5': 'Communication about medicines',
        'H_COMP_6': 'Discharge information',
        'H_COMP_7': 'Care transition',
        'H_CLEAN_HSP': 'Cleanliness of hospital environment',
        'H_QUIET_HSP': 'Quietness of hospital environment',
        'H_HSP_RATING': 'Hospital rating (0-10)',
        'H_RECMND': 'Willingness to recommend hospital'
    }
    
    for code, label in mapping.items():
        if measure_id.startswith(code):
            suffix = ''
            if measure_id.endswith('_A_P'):
                suffix = ' (always)'
            elif measure_id.endswith('_SN_P'):
                suffix = ' (sometimes/never)'
            elif measure_id.endswith('_U_P'):
                suffix = ' (usually)'
            elif measure_id.endswith('_Y_P'):
                suffix = ' (yes)'
            elif measure_id.endswith('_DY'):
                suffix = ' (definitely yes)'
            elif measure_id.endswith('_9_10'):
                suffix = ' (9-10 rating)'
            elif measure_id.endswith('_8_9_10'):
                suffix = ' (8-10 rating)'
            elif measure_id.endswith('_6_7_8_9_10'):
                suffix = ' (6-10 rating)'
            return label + suffix
    
    return measure_id

print("✅ Helper functions defined")
```

## 🔍 Data Filtering

```python
# Filter for target hospitals
df_filtered = df_hcahps[df_hcahps['Facility Name'].isin(HOSPITALS)].copy()

# Apply measure name cleaning
df_filtered['Measure_pretty'] = df_filtered['Measure ID'].apply(clean_measure_name)

print(f"✅ Filtered data: {len(df_filtered)} rows")
print(f"🏥 Hospitals in dataset: {df_filtered['Facility Name'].nunique()}")
```
